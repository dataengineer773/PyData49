```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({'key': ['a', 'b', 'c'] * 4,
                   'value': np.arange(12.)})
df
```

```python
g = df.groupby('key')['value']
g.mean()
```

```python
def get_mean(group):
    return group.mean()
g.transform(get_mean)
```

```python
g.transform('mean')
```

```python
def times_two(group):
    return group * 2
g.transform(times_two)
```

```python
def get_ranks(group):
    return group.rank(ascending=False)
g.transform(get_ranks)
```

```python
def normalize(x):
    return (x - x.mean()) / x.std()
```

```python
g.transform(normalize)
g.apply(normalize)
```

```python
g.transform('mean')
normalized = (df['value'] - g.transform('mean')) / g.transform('std')
normalized
```
